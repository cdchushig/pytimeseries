## Visualization of time series

```python
plot_time_series(v_ts_w1, title='ts_w1')
```

- The data consists of 45 daily observations of consecutive days between November 4 to January 10.
- The figure suggests a **stationary process** with constant mean and variance.
- The sample ACF decays exponentially and the sample PACF has a single significant spike at lag 1, indicating that time series **is likely** to be generated by a AR(1) process.

```python
plot_acf(v_ts_w1, lags=20, alpha=0.05)
plt.show()
```

```python
plot_pacf(v_ts_w1, lags=20, alpha=0.05, method='ywm')
plt.show()
```
